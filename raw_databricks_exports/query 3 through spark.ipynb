```python
# Load the Spark DataFrame
df = spark.table("default.ecommerce_customers")

# Apply the filter and compute average
df_filtered = df.filter(df["`Avg. Session Length`"] > 34)

# Compute the average of "Yearly Amount Spent"
df_filtered.agg({"Yearly Amount Spent": "avg"}).withColumnRenamed("avg(Yearly Amount Spent)", "avg_spending_long_sessions").show()
```
