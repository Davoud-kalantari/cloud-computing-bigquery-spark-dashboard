# Query 1 in Spark SQL

Purpose: compute the average `Yearly Amount Spent` from the Databricks table `default.ecommerce_customers`.

Why this notebook is kept: the course required the query to be shown in Spark SQL form in addition to BigQuery and Spark DataFrame/operator form.


```python
%sql
SELECT
  AVG(`Yearly Amount Spent`) AS avg_spending
FROM default.ecommerce_customers
```
