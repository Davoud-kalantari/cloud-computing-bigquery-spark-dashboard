# Query 3 in Spark SQL

Purpose: compute average `Yearly Amount Spent` for rows where `Avg. Session Length > 34`.

Why this notebook is kept: the course required a Spark SQL implementation for the third Databricks query.

Important note: with the currently available evidence, this Query 3 does not match the Query 3 preserved in the cleaned Google Cloud section, where Query 3 is `top 5 customers by yearly spending`. The numbering is preserved from the Databricks exports and the mismatch is documented honestly.


```python
%sql
SELECT
  AVG(`Yearly Amount Spent`) AS avg_spending_long_sessions
FROM default.ecommerce_customers
WHERE `Avg. Session Length` > 34
```
