# Query 2 in Spark SQL

Purpose: compare average `Time on App` and average `Time on Website` using the Databricks SQL execution mode.

Why this notebook is kept: the course required a Spark SQL version of the same analytical query alongside the BigQuery and Spark DataFrame/operator implementations.


```python
%sql
SELECT
  AVG(`Time on App`) AS avg_time_on_app,
  AVG(`Time on Website`) AS avg_time_on_website
FROM default.ecommerce_customers
```
