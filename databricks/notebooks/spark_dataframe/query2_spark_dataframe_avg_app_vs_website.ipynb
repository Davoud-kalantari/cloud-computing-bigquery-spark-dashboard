# Query 2 with Spark DataFrame operators

Purpose: compute average `Time on App` and average `Time on Website` using Spark DataFrame expressions.

Why this notebook is kept: the course required the query to be shown in Spark DataFrame/operator form in addition to BigQuery and Spark SQL.


```python
df = spark.table("default.ecommerce_customers")

df.selectExpr(
    "AVG(`Time on App`) as avg_time_on_app",
    "AVG(`Time on Website`) as avg_time_on_website"
).show()
```
