# Query 1 with Spark DataFrame operators

Purpose: compute the average `Yearly Amount Spent` using Spark DataFrame loading plus SQL over a temporary view.

Why this notebook is kept: the course required a Spark operator/DataFrame version in addition to the BigQuery and Spark SQL versions.


```python
df = spark.read.table("default.ecommerce_customers")
df.createOrReplaceTempView("ecommerce")

query1 = spark.sql("""
    SELECT AVG(`Yearly Amount Spent`) AS avg_spending
    FROM ecommerce
""")

query1.show()
```
