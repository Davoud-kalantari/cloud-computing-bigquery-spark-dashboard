# Query 3 with Spark DataFrame operators

Purpose: compute average `Yearly Amount Spent` after filtering rows where `Avg. Session Length > 34`.

Why this notebook is kept: the course required a Spark DataFrame/operator implementation of the third Databricks query.

Important note: this Query 3 is preserved exactly as evidenced in the Databricks exports. It does not currently match the Query 3 preserved in the cleaned Google Cloud section, and that mismatch is documented in the mapping notes.


```python
df = spark.table("default.ecommerce_customers")

df_filtered = df.filter(df["`Avg. Session Length`"] > 34)

df_filtered.agg({"Yearly Amount Spent": "avg"}) \
    .withColumnRenamed("avg(Yearly Amount Spent)", "avg_spending_long_sessions") \
    .show()
```
