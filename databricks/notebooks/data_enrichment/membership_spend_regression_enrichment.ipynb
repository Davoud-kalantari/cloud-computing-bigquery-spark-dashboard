# Data Enrichment: membership vs spending regression

Purpose: explore the relationship between `Length of Membership` and `Yearly Amount Spent` using the same Databricks customer table.

Honesty note: this notebook is kept as a data enrichment example, not as a Spark MLlib notebook. The current implementation uses Spark for filtering and column selection, then converts to pandas and applies NumPy regression plus matplotlib plotting.


```python
from pyspark.sql.functions import col

df = spark.table("default.ecommerce_customers")

filtered_df = df.filter(
    col("Length of Membership").isNotNull() &
    col("Yearly Amount Spent").isNotNull()
)

pandas_df = filtered_df.select(
    "Length of Membership", "Yearly Amount Spent"
).toPandas()

pandas_df = pandas_df.dropna()

import numpy as np
import matplotlib.pyplot as plt

x = pandas_df["Length of Membership"]
y = pandas_df["Yearly Amount Spent"]

a, b = np.polyfit(x, y, 1)

plt.scatter(x, y, label='Data points')
plt.plot(x, a * x + b, color='red', label=f'Regression line: y = {a:.2f}x + {b:.2f}')
plt.xlabel("Length of Membership")
plt.ylabel("Yearly Amount Spent")
plt.legend()
plt.show()
```
