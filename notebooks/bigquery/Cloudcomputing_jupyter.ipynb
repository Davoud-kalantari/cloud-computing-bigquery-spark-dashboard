```python
from google.cloud import bigquery
import os
import pandas as pd

credential_path = os.environ.get("GOOGLE_APPLICATION_CREDENTIALS")
if not credential_path:
    raise ValueError("Set GOOGLE_APPLICATION_CREDENTIALS to a local service-account JSON file before running this notebook.")

client = bigquery.Client.from_service_account_json(credential_path)
print("Connected to project:", client.project)
```

```python
# query 1 result 
query1 = """
SELECT AVG(`Yearly Amount Spent`) AS media_spesa_annua
FROM `ccbm-exam-dk-2025.ecommerce_data.customers`
"""

df1 = client.query(query1).to_dataframe()
df1
```

```python
df1.to_csv("avg_yearly_spending.csv", index=False)

```

```python
# query 1 code to save on bucket
from google.cloud import storage
import os

# (if not already set)
# Credentials are expected from the local GOOGLE_APPLICATION_CREDENTIALS environment variable

# Upload to GCS
client = storage.Client()
bucket_name = "ccbm-exam-dk-2025"
destination_blob_name = "avg_yearly_spending.csv"
source_file_name = "avg_yearly_spending.csv"

bucket = client.bucket(bucket_name)
blob = bucket.blob(destination_blob_name)
blob.upload_from_filename(source_file_name)

print("✅ File 'avg_yearly_spending.csv' uploaded successfully to GCS.")
```

```python
from google.cloud import bigquery

# Initialize the correct BigQuery client
client = bigquery.Client()

# Your query
query2 = """
SELECT 
    AVG(`Time on App`) AS media_tempo_app,
    AVG(`Time on Website`) AS media_tempo_web
FROM 
    `ccbm-exam-dk-2025.ecommerce_data.customers`
"""

# Execute query and save to DataFrame
df2 = client.query(query2).to_dataframe()

# Show result
df2
```

```python
# Ensure you're still using the correct BigQuery client
client = bigquery.Client()

# Third query: Top 5 customers by highest spending
query3 = """
SELECT 
    `Email`, 
    `Yearly Amount Spent`
FROM 
    `ccbm-exam-dk-2025.ecommerce_data.customers`
ORDER BY 
    `Yearly Amount Spent` DESC
LIMIT 5
"""

# Execute and convert to DataFrame
df3 = client.query(query3).to_dataframe()

# Show the result
df3
```

```python
from google.cloud import storage
import os

# Save the query result to a local CSV file
df3.to_csv("top_5_customers.csv", index=False)

# Set your Google Cloud credentials if not already set
# Credentials are expected from the local GOOGLE_APPLICATION_CREDENTIALS environment variable

# Create a storage client
client_storage = storage.Client()

# Define the bucket name and destination file name
bucket_name = "ccbm-exam-dk-2025"
destination_blob_name = "top_5_customers.csv"
source_file_name = "top_5_customers.csv"

# Upload the file
bucket = client_storage.bucket(bucket_name)
blob = bucket.blob(destination_blob_name)
blob.upload_from_filename(source_file_name)

print("✅ 'top_5_customers.csv' successfully uploaded to GCS bucket.")
```
